```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
```

```python
text="Shakespeare was a great writer and poet."
```

```python
chars=sorted(list(set(text)))
char_to_index={c: i for i, c in enumerate(chars)}
index_to_char={i: c for i,c in enumerate(chars)}
```

```python
sequence_length=5
X=[]
y=[]

for i in range(len(text)- sequence_length):
    seq=text[i:i+sequence_length]
    target=text[i+sequence_length]

    X.append([char_to_index[c] for c in seq])
    y.append(char_to_index[target])

X=np.array(X)
y=np.array(y)

X=X/float(len(chars))
```

```python
# Step 4: Build model
model = Sequential([
    LSTM(128, input_shape=(sequence_length, 1)),
    Dense(len(chars), activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy', optimizer='adam')

# Reshape X for LSTM
X = np.reshape(X, (X.shape[0], X.shape[1], 1))
```

```python
# Step 5: Train model
model.fit(X, y, epochs=30, batch_size=32)
```

```python
# Step 6: Generate text
start = "Shake"
generated = start

for _ in range(20):
    pattern = [char_to_index[c] for c in start]
    pattern = np.reshape(pattern, (1, sequence_length, 1))
    pattern = pattern / float(len(chars))
    
    prediction = model.predict(pattern, verbose=0)
    index = np.argmax(prediction)
    result = index_to_char[index]
    
    generated += result
    start = start[1:] + result

print("Generated Text:", generated)
```
